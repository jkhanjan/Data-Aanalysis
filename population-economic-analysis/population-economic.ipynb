```python
# Sorts regions automatically from highest average stress to lowest
order = df.groupby('region')['final_economic_stress_score'].mean().sort_values(ascending=False).index

sns.barplot(
    data=df, 
    x='final_economic_stress_score', 
    y='region', 
    order=order
)
```

```python
components = [
    'inflation_score', 'unemployment_score', 'gdp_growth_score', 
    'income_vulnerability_score', 'food_pressure_score'
]

# Get average of each component per region
region_drivers = df.groupby('region')[components].mean()

sns.heatmap(region_drivers, annot=True, cmap='coolwarm')
```

```python
order = df.groupby('income_group')['final_economic_stress_score'].mean().sort_values(ascending=False).index

sns.barplot(
    data=df, 
    x='final_economic_stress_score', 
    y='income_group', 
    order=order,
    errorbar=None
)
```

```python
df.groupby('year')['final_economic_stress_score'].mean()
sns.lineplot(
    data=df, 
    x='year', 
    y='final_economic_stress_score', 
    marker='o'
)
```

```python
cols = ['inflation_score', 'unemployment_score', 'gdp_growth_score', 
        'income_vulnerability_score', 'food_pressure_score', 'final_economic_stress_score']

sns.heatmap(df[cols].corr(), annot=True, cmap='coolwarm', fmt=".2f")

correlations = df[cols].corr()['final_economic_stress_score'].abs().sort_values(ascending=False)
print(correlations)
```
